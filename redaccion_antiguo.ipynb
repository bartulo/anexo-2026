## EDAFOLOGÍA

El análisis de la edafología es clave para entender el comportamiento del fuego, ya que determina la humedad del terreno, el tipo de vegetación (combustible) que crece en él y la accesibilidad para los equipos de extinción, puesto que la calidad de las pistas forestales dependerá del tipo de suelos sobre los que se asienten. Estas pistas constituyen una estructura esencial en la prevención y lucha contra incendios forestales como se verá en el apartado 3.

Se muestra mapa edafológico del municipio.

Según la vegetación potencial, capacidad de retención del agua y recuperación post incendio se han clasificado 5 niveles de riesgo:

*   **Zona de Riesgo Alto (Vertientes Áridas y Suelos Erosionables):** 
    Comprende laderas de fuerte pendiente, escarpes y lomas sometidos a procesos activos de erosión sobre materiales de margas y areniscas. Son suelos delgados ("esqueléticos") con nula capacidad de retención hídrica, lo que somete a la vegetación (matorral mediterráneo y pinar) a un estrés hídrico extremo en verano. Topográficamente, sus pendientes favorecen el "efecto chimenea", acelerando la velocidad de propagación del frente de llama. Además, presentan una accesibilidad terrestre muy compleja para los equipos de extinción y un riesgo crítico de pérdida irreversible de suelo por erosión hídrica post-incendio.

*   **Zona de Riesgo Moderado-Alto (Glacis y Terrazas Secas):** 
    Formada por plataformas llanas o suavemente inclinadas, relieves de glacis y terrazas fluviales medias y altas. Aunque topográficamente facilitan el tránsito de vehículos, edafológicamente destacan por su carácter pedregoso o por la presencia de horizontes cementados (costras calizas). Esto limita la profundidad de las raíces y provoca una rápida desecación del combustible fino (pastizales y cultivos de secano) al inicio del estío, convirtiéndose en zonas propensas a incendios muy rápidos y guiados por el viento.

*   **Zona de Transición y Cultivo (Laderas de Acumulación y Vaguadas Altas):** 
    Esta zona agrupa suelos profundos y coluviales en bases de ladera y vaguadas, desarrollados sobre arcillas, limos y margas. Al acumular nutrientes y mantener una humedad edáfica moderada durante la primera mitad del verano, sostienen una alta productividad biológica (masas forestales densas, matorral espeso y cultivos de cereal). Su riesgo es estacionalmente tardío: si el incendio ocurre a finales de verano bajo condiciones de sequía prolongada, la gran cantidad de biomasa disponible (combustible pesado) puede generar incendios de gran carga térmica y alta intensidad.

*   **Zona de Riesgo Bajo / Barrera Natural (Suelos Fluviales e Húmedos):** 
    Constituida por los fondos de barranco, llanuras aluviales, masas de agua y terrazas bajas asociadas a las cuencas de los ríos Aragón, Irati, Areta y Salazar. Estos suelos disfrutan de una capa freática accesible y alta humedad constante, lo que permite el desarrollo de vegetación riparia (bosques de ribera y huertas) con un elevado contenido de agua en sus tejidos. Actúan como cortafuegos naturales capaces de frenar o amortiguar el avance del fuego, además de albergar los puntos clave para el abastecimiento hídrico de los medios aéreos.

*   **Área Antrópica (Suelos Urbanizados e Infraestructuras):** 
    Incluye los cascos urbanos, la red de carreteras, canteras y vertederos. Desde la perspectiva del comportamiento del fuego, las zonas extractivas e infraestructuras viales carecen de combustible vegetal, funcionando como zonas de oportunidad para el despliegue de líneas de defensa terrestres. El foco de gestión en esta zona no es la propagación forestal, sino la seguridad humana ante el riesgo crítico que supone la interfaz urbano-forestal en los núcleos de población de Sangüesa y Aibar.

## TOPOGRAFÍA

La topografía de un terreno es una característica esencial para determinar la evolución de un incendio forestal. Junto al tiempo atmosférico y al combustible forman el triangulo de comportamiento del fuego. 

La topografía de los municipios de Sangüesa y Áibar se caracteriza por ser una zona de transición entre el Pirineo y la depresión del Ebro. En el sector occidental del área de estudio se suceden pequeñas cadenas montañosas orientadas de oeste a este. Por el contrario, en la zona oriental, el río Aragón rompe la continuidad de estos relieves, abriendo un amplio valle con orientación norte-sur.

En consecuencia, el comportamiento previsto de un incendio variará sustancialmente entre ambas zonas. Mientras que en el sector oeste el fuego mostrará un comportamiento cambiante debido a factores orográficos locales —como vientos fuertes en las crestas, efecto Venturi en los fondos de valle o contravientos en las laderas a sotavento—, en el valle del río Aragón predominará un incendio conducido por el viento, ya que los dos vientos dominantes (el cierzo del norte y el bochorno del sur) se pueden llegar a alinear con la dirección del eje fluvial.

Se muestra mapa con las especies forestales en la zona de estudio:

```python
img = Image.open('FINALES/IMAGENES/forestal.png')
display(img, metadata={'pie': 'Mapa especies forestales. Fuente: elaboración propia a partir de datos obtenidos en el IDENA'})
```

La vegetación forestal de la zona de estudio está dominada por especies mediterráneas. No obstante, existe una notable heterogeneidad en sus masas forestales, las cuales pueden clasificarse según su piroecología y respuesta adaptativa ante los incendios forestales:

* **Vegetación de ribera (Alta vulnerabilidad, baja peligrosidad):** Debido a las características hídricas y edáficas de su entorno, estas masas presentan un riesgo de ignición muy bajo, al asentarse sobre suelos con una capa freática elevada y próximos a cauces fluviales. Sin embargo, en escenarios de sequías estivales extremas, un eventual incendio resultaría altamente devastador, ya que estas comunidades carecen de estrategias evolutivas de colonización o regeneración post-fuego.
* **Matorral mediterráneo (Alta peligrosidad, baja vulnerabilidad individual):** Formado principalmente por comunidades de coscoja (*Quercus coccifera*), este estrato presenta una elevada inflamabilidad. Pese a ello, los individuos persisten con éxito gracias a su potente capacidad rebrotadora, lo que garantiza una rápida recuperación tras el paso del fuego.
* **Pinares de pino laricio o negral (*Pinus nigra*) (Alta vulnerabilidad, peligrosidad media):** Se trata de una especie adaptada evolutivamente a regímenes de incendios superficiales de baja intensidad, gracias al aislamiento térmico de su gruesa corteza basal y a la autopoda natural que eleva sus copas. No obstante, ante fuegos de copa o de alta severidad, carece de mecanismos de regeneración post-fuego (como la serotinia), lo que compromete gravemente la persistencia de la masa. En el área de estudio, estas formaciones se localizan preferentemente como repoblaciones en las laderas de umbría (norte).
* **Pinares de pino carrasco (*Pinus halepensis*) (Baja vulnerabilidad específica, alta vulnerabilidad individual, alta peligrosidad):** Estructuralmente, estas masas tienden a propagar fuegos de alta intensidad debido a la arquitectura de su copa y a la continuidad vertical del combustible. Aunque el incendio destruye al individuo adulto, la especie asegura su persistencia poblacional mediante bancos de semillas aéreos (piñas serótinas) que se abren masivamente tras el fuego, colonizando el suelo desnudo con una mínima competencia biológica. Conviene tener en cuenta esta característica en la planificación de quemas prescritas. Una intensidad moderada activa la apertura de las piñas y genera una densidad alta de brinzales, logrando un efecto contrario al deseado. En el territorio analizado, las masa de pino carrasco ocupan predominantemente las laderas de solana (sur).
* **Masas de frondosas —*Quercus faginea* y *Quercus humilis*— (Baja vulnerabilidad, baja peligrosidad):** Situadas en los fondos de valle del área de estudio, están constituidas por ejemplares de gran porte que generan un dosel cerrado. Esta densa sombra reduce el desarrollo del sotobosque y mantiene un microclima húmedo que dificulta la ignición. En caso de verse afectadas por el fuego, ambas quercíneas poseen una excelente resiliencia basada en el rebrote, lo que les confiere una clara ventaja competitiva durante la sucesión post-incendio.

Para caracterizar los incendios históricos de los municipios de Sangüesa y Aibar se han usado dos fuentes de datos:

## IDENTIFICACIÓN DE LAS ACTIVIDADES O USOS SUSCEPTIBLES DE PRODUCIR INCENDIOS

Para identificar aquellas actividades o usos que pueden producir incendios se va a recurrir nuevamente a la Estadística General de Incendios Forestales. A continuación se muestra una tabla con los incendios y conatos producidos en los municipios de Sangüesa y Aibar en el periodo 1968-2021 agrupados por sus causas.

```python
causalidad = incendios_egif.groupby('Causa').agg(
    num_incendios = ('Inicio', 'count'),
    sup_quemada = ('Superficie', 'sum')
).reset_index().sort_values('sup_quemada', ascending=False)
```

```python
causalidad.rename(columns={
    'num_incendios': 'Núm. de incendios',
    'sup_quemada': 'Superficie'
}, inplace=True)
```

```python
display(Markdown(causalidad.to_latex(float_format='%.2f', index=False)))
```

En las estadísticas EGIF viene reflejada la motivación de los incendios intencionados. Se procede a agrupar los 10 incendios intencionados en la zona de estudio por su motivación.

```python
incendios_intencionados = incendios_egif[incendios_egif['Causa'] == 'Incendio Intencionado']
```

```python
intencionados_agrupados = incendios_intencionados.groupby('Motivación').agg(
    num_incendios = ('Inicio', 'count'),
    sup_quemada = ('Superficie', 'sum')
).reset_index().sort_values('sup_quemada', ascending=False)
```

```python
intencionados_agrupados.rename(columns={
    'num_incendios': 'Núm. de incendios',
    'sup_quemada': 'Superficie'
}, inplace=True)
```

```python
display(Markdown(intencionados_agrupados.to_latex(float_format='%.2f', index=False)))
```

En términos de superficie afectada, las principales causas de los incendios son las actividades agrícolas, seguidas de las ganaderas y, en tercera posición, los accidentes en líneas eléctricas.

## MEDIOS MITECO

Si un incendio forestal pasara a **Situación Operativa 2**, la Comunidad Foral de Navarra podría solicitar el apoyo de los recursos estatales que el **Ministerio para la Transición Ecológica y el Reto Demográfico (MITECO)** pone a disposición de las comunidades autónomas. 

Los medios de titularidad estatal desplegados de forma más próxima a la Comunidad Foral son los siguientes:

*   **BRIF (Brigadas de Refuerzo en Incendios Forestales):** Equipos de extinción helitransportados altamente especializados. Cada brigada está compuesta por 16 bomberos forestales, 2 capataces y 1 técnico, y cuenta con dos helicópteros de tipo medio (MIKE) para el transporte de personal y el bombardeo de agua. Las bases más cercanas a Navarra son las de **Lubia (Soria)** y **Daroca (Zaragoza)**.
*   **Aviones de Carga en Tierra (ACT o TANGO):** Aeronaves de ala fija que cargan agua en pistas de aterrizaje y disponen de sistemas para mezclar el agua con retardante químico. Durante la campaña estival de alto riesgo, opera **un avión ACT desde el Aeropuerto de Pamplona-Noáin** y **dos desde el Aeropuerto de Logroño-Agoncillo (La Rioja)**.
*   **Helicópteros ligeros de bombardeo (LIMA):** Helicópteros de tipo 3 con una capacidad de descarga inferior a los 1.000 litros. Están diseñados para el ataque rápido y directo al fuego mediante descargas de agua, y no para el transporte de brigadas. 
*   **Helicópteros medios de bombardeo (MIKE):** Helicópteros de tipo 2 con una capacidad de descarga de unos 1.200 litros. Al igual que los LIMA, operan principalmente como bombarderos en misiones de extinción. El MITECO posiciona dos helicópteros MIKE en la base de **Plasencia del Monte (Huesca)**.
*   **Aviones anfibios de gran capacidad (FOCA):** Aeronaves de ala fija con una capacidad de carga de entre 5.500 y 6.100 litros, capaces de captar agua directamente en embalses o en el mar. En la **Base Aérea de Zaragoza** se destaca un avión FOCA operado por el 43 Grupo del Ejército del Aire y del Espacio, bajo la dirección funcional del MITECO.

# IDENTIFICACIÓN DE PUNTOS ESTRATÉGICOS DE GESTIÓN

Para la identificación de los Puntos Estratégicos de Gestión (PEG) se ha empleado una **metodología mixta** que combina la **simulación computacional** con el **criterio experto**. Por un lado, se utilizó el software *FlamMap* para ejecutar simulaciones que permiten localizar los puntos donde el comportamiento del fuego propicia la ramificación y el incremento del potencial de propagación. Por otro lado, mediante criterio experto, se evaluó la viabilidad e idoneidad técnica de ejecutar intervenciones en los emplazamientos preseleccionados.

El modelo de simulación en *FlamMap* requiere como variables de entrada (*inputs*) cartografía topográfica y de combustibles (pendientes, orientaciones y modelos de combustible), así como variables meteorológicas críticas (temperatura, velocidad del viento y humedad relativa). 

La caracterización meteorológica introducida en el modelo se fundamenta en el análisis histórico de incendios de la Comunidad Foral de Navarra (detallado en el apartado 2.1). Dicho estudio determinó que el escenario sinóptico más desfavorable corresponde a una situación de **componente continental sahariana acompañada de viento sur**.

Bajo estas condiciones críticas, se ejecutaron **cuatro escenarios de simulación** variando el foco de inicio del incendio. Los puntos de ignición se ubicaron en zonas de alta probabilidad de ocurrencia (vías de comunicación, interfases agrícolas, entre otros) y en posiciones estratégicas capaces de desarrollar incendios de gran magnitud que afecten de forma directa a las masas forestales objeto de ordenación.

Como resultado de este proceso de análisis y selección, se han diseñado un total de **9 PEG**, los cuales se describen a continuación:


```python
img = Image.open('FINALES/IMAGENES/peg.png')

display(img, metadata={'pie': 'Mapa de localización de los Puntos Estratégicos de Gestion propuestos. Se muestran cierres ganaderos y delimitación de los montes comunales.'})
```

Los PEG propuestos se clasifican en tres tipologías estructurales:

* **Defensivos**: Orientados a la protección de infraestructuras y bienes materiales en la interfase urbano-forestal. Su objetivo es reducir la carga de combustible en áreas perimetrales próximas a núcleos de población.
* **En nudos de barranco**: Diseñados para mitigar el comportamiento del fuego en zonas topográficamente complejas, evitando la ramificación y aceleración de la propagación del frente.
* **En cumbreras**: Enfocados en reducir la intensidad lineal del incendio mediante la creación de puntos de oportunidad tácticos, limitando la progresión vertical de las llamas y minimizando el riesgo de propagación por paveseo.

Para cada uno de los PEG propuestos se ha elaborado una ficha técnica descriptiva que detalla su superficie cartográfica, la justificación operativa de su selección y las propuestas de actuación correspondientes.
